```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('data-export (1).csv')
```

```python
df.head()
```

```python
df.columns=df.iloc[0]
df.head()
```

```python
# Droping the 0th row and renaming the column names
df = df.drop(index=0).reset_index(drop=True)
df.columns = ['channel group','DateHour','Users','Sessions','Engaged Sessions','Average engagement time per session','Engaged sessions per user','Events per session','Engagement rate','Event count']
```

```python
df.head()
```

```python
df.info()
```

# Cleaning Data and Data Validation for above data set

```python
# converting the DateHour column from object to datetime
df['DateHour'] = pd.to_datetime(df['DateHour'], format='%Y%m%d%H',errors='coerce')
```

```python
df.info()
```

```python
df.head()
```

```python
# converting other columns from object to numeric
numeric_cols = df.columns.drop(['channel group','DateHour'])
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric,errors='coerce')
df['Hour'] = df['DateHour'].dt.hour
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

## What patterns and trends can you observe in website Sessions and user over time?

```python
sns.set(style="whitegrid")
```

```python
plt.figure(figsize=(10,5))
df.groupby("DateHour")[["Sessions","Users"]].sum().plot(ax=plt.gca())
plt.title("Sessions and users over time")
plt.xlabel("DateHour")
plt.ylabel("count")
plt.show()
```

Conclusion:-
People are sessioning in the website more between 16 to 20

## Which marketing channel brought the highest number of users to the website?

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="channel group", y="Users", estimator=np.sum, palette="viridis")
plt.title("👥 Total Users by Channel")
plt.xticks(rotation=45)
plt.show()
```

Coclusion:- The Organic social channel has more users and they have to give more focus to Unassigned,organic video and Email channels

```python
df.head()
```

## Which channel has the highest Average engagement time by channel?

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="channel group", y="Average engagement time per session",estimator=np.mean,palette='magma')
plt.title("Avg Engagement Time by Channel")
plt.xticks(rotation=45)
plt.show()
```

```python
avg_engagement = df.groupby("channel group")["Average engagement time per session"].mean()
plt.figure(figsize=(8, 5))
plt.pie(avg_engagement,labels=avg_engagement.index,autopct='%1.1f%%',startangle=90)
plt.title("Avg Engagement Time by Channel")
plt.axis('equal') 
plt.show()
```

Conclusion:- users are engaged in Organic Video Channel and if website work more on Email channel then there is chances that Email channel can grow

## How does the Engagement Rate  vary across different channel?

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="channel group", y="Engagement rate", palette="coolwarm")
plt.title("📦 Engagement Rate Distribution by Channel")
plt.xticks(rotation=45)
plt.show()
```

Conclusion:- Referral has most engagement rate and Email has lowest engagement rate so, Refferal is high performer

## Which channels are driving more engaged sessions as compared to non-engaged ones?

```python
df.head()
```

```python
session_df = df.groupby("channel group")[["Sessions", "Engaged Sessions"]].sum().reset_index()
session_df["Non-Engaged"] = session_df["Sessions"] - session_df["Engaged Sessions"]
session_df_melted = session_df.melt(id_vars="channel group", value_vars=["Engaged Sessions",'Non-Engaged'])
plt.figure(figsize=(8, 5))
sns.barplot(data=session_df_melted, x="channel group", y="value", hue="variable")
plt.title(" Engaged vs Non-Engaged Sessions")
plt.xticks(rotation=45)
plt.show()
```

Conclusion:

- The Organic Social has suppose 30200 engaged sessions but it also has approximetly 28000 non-engaged sessions and 
- Other side we can see Direct channel has more Non-engaged sessions than Engagaed sessions so company have to work on it

## At what hours of the day does each channel drive the most traffic?


```python
heatmap_data = df.groupby(["Hour", "channel group"])["Sessions"].sum().unstack()
plt.figure(figsize=(12, 6))
sns.heatmap(heatmap_data, cmap="YlGnBu", linewidths=.5, annot=True, fmt='.0f')
plt.title("⏰ Traffic by Hour and Channel")
plt.xlabel("Channel Group")
plt.ylabel("Hour of Day")
plt.show()
```

Conclusion:-
- we can see from the above graph that in Organic Social 0 means at 12AM there is most traffic
-  In referrel we  can see that around 11AM there is most traffic

## What is engagement rate vs sessions over time?

```python
df.head()
```

```python
df_plot = df.groupby("DateHour")[["Engagement rate", "Sessions"]].mean().reset_index()
plt.figure(figsize=(10, 5))
plt.plot(df_plot["DateHour"], df_plot["Engagement rate"], label="Engagement rate", color='green')
plt.plot(df_plot["DateHour"], df_plot["Sessions"], label="Sessions", color="blue")
plt.title("📊 Engagement Rate vs Sessions Over Time")
plt.xlabel("DateHour")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.show()
```

Conclusion:-
- Here we can see the vertical line for Engagement rate because from 5th April to 5th May users are sessioning but not engaging in the website
